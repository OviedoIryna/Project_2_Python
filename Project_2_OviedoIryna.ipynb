**Tasc 5. Визначення популярності Python**

```python
#Tasc 5
df['LanguageHaveWorkedWith']
```

```python
# 5.1 Створюємо новий стовпець UsersPython в df
#застосовуємо метод contains з параметрами case=False не враховує регістр написання Python,
#na=False якщо значення = NaN → ставить False
df['UsersPython'] = df['LanguageHaveWorkedWith'].str.contains('Python', case=False, na=False)
```

```python
#5.2 Рахуємо кількість респондентів, які використовують Python
python_count = df['UsersPython'].sum()
total = len(df)

#5.3 У відсотках кількість респондентів, які використовують Python
python_percentage = round((python_count / total) * 100)

print(f'Кількість респондентів, які використовують Python : {python_count}')
print(f'Відсоток популярності Python : {python_percentage}%')
```

**Tasc 7. Географічний аналіз компенсації Python-розробників**

```python
#Tasc 7
# 7.1 Відбираємо (фільтруємо) вже тиз респондентів, хто працює з Python та має НЕ порожні зарплати
python_df = df[df['UsersPython'] & df['ConvertedCompYearly'].notna()]

# 7.2 групування по країнах значення компенсації (зарплати)
salary_by_country = python_df.groupby('Country')['ConvertedCompYearly'].agg(['mean', 'median'])

# 7.3. округлення
salary_by_country = salary_by_country.round(0)

# 7.4 сортування
salary_by_country.sort_values(by='mean', ascending=False).head(10)
```

**Tasc 8. Аналіз освіти найбільш оплачуваних спеціалістів**

```python
#Tasc 8
# 8.1 прибираємо пропуски по зарплаті
df_clean = df.dropna(subset=['ConvertedCompYearly'])

# 8.2 сортуємо за зарплатою (від більшої до меншої) та лишаємо топ5 зарплат
top_5 = df_clean.sort_values(by='ConvertedCompYearly', ascending=False).head(5).reset_index(drop=True)

# 8.3 дивимось зарплату та освіту в табличному вигляді та оновлюємо індекси, видаляємо "старі" індекси
top_5[['ConvertedCompYearly', 'EdLevel']]
```

**Tasc 9. Аналіз популярності Python по віковим категоріям**

```python
#Tasc 9
# 9.1 групування по вікових категоріях та визначення відсотка використання Python
python_by_age = df.groupby('Age')['UsersPython'].mean() * 100

# 9.2 округлення
python_by_age = python_by_age.round(2)

# 9.3 сортування та даємо назву стовпцю з відсотками
python_by_age.sort_values(ascending=False)
python_by_age = python_by_age.reset_index(name='PythonPercentage')
python_by_age
```

**Tasc 10. Аналіз індустрій серед високооплачуваних віддалених працівників**

```python
# 10.1 прибираємо NaN по зарплаті
df_clean = df.dropna(subset=['ConvertedCompYearly'])

# 10.2 знаходимо 75-й перцентиль
percentile_75 = df_clean['ConvertedCompYearly'].quantile(0.75)

# 10.3 фільтруємо:
high_paid_remote = df_clean[
    (df_clean['ConvertedCompYearly'] > percentile_75) &
    (df_clean['RemoteWork'].str.contains('remote', case=False, na=False))
]

# 10.4 дивимось найпопулярніші індустрії
top_industries = high_paid_remote['Industry'].value_counts().head(10).reset_index(name='Count')

top_industries
```

```python
# Візуалізація у вигляді горизонтальної гістограми
plt.figure()

plt.barh(top_industries['Industry'], top_industries['Count'])

plt.title('Top Industries Among High-Paid Remote Workers')
plt.xlabel('Count of Respondents')

plt.tight_layout()
plt.show()
```
